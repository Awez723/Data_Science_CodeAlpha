```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('Unemployment in India.csv')
print("Data loaded successfully.")
print(df.head())
print("Row & Columns:", df.shape)
print(df.columns)
print(df.isnull().sum())
df.columns = df.columns.str.strip()
print(df.columns)
df["Date"] = pd.to_datetime(df["Date"])
print(df["Date"].head())
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
print(df["Date"].head())
print(df.describe())
average_unemployment = df["Estimated Unemployment Rate (%)"].mean()
print("Average Unemployment Rate:", average_unemployment, "%")
region_unemployment = df.groupby("Region")["Estimated Unemployment Rate (%)"].mean()
print(region_unemployment)
monthly_unemployment = df.groupby("Date")["Estimated Unemployment Rate (%)"].mean()

plt.figure(figsize=(12, 6))
plt.plot(monthly_unemployment.index, monthly_unemployment.values)

plt.title("Unemployment Rate Trend in India")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()


covid_data = df[
    (df["Date"] >= "2020-03-01") &
    (df["Date"] <= "2020-12-31")
]
print("COVID-19 Period Data:")
print(covid_data[["Date", "Estimated Unemployment Rate (%)"]].head())
print("\nAverage Unemployment Rate during COVID-19:")
print(covid_data["Estimated Unemployment Rate (%)"].mean(), "%")

plt.figure(figsize=(12, 6))

plt.plot(
    covid_data["Date"],
    covid_data["Estimated Unemployment Rate (%)"],
    marker="o"
)

plt.title("Unemployment Rate During COVID-19")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()

area_unemployment = df.groupby("Area")["Estimated Unemployment Rate (%)"].mean()
print("Average Unemployment Rate by Area:")
print(area_unemployment)

area_unemployment.plot(kind="bar", figsize=(8, 5))

plt.title("Average Unemployment Rate: Rural vs Urban")
plt.xlabel("Area")
plt.ylabel("Unemployment Rate (%)")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()

plt.figure(figsize=(10, 6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()

highest_unemployment = df["Estimated Unemployment Rate (%)"].max()
print("Highest Unemployment Rate:", highest_unemployment, "%")

lowest_unemployment = df["Estimated Unemployment Rate (%)"].min()
print("Lowest Unemployment Rate:", lowest_unemployment, "%")

highest_region = region_unemployment.idxmax()
highest_rate = region_unemployment.max()

print("Region with Highest Unemployment Rate:", highest_region)
print("Unemployment Rate:", highest_rate, "%")

plt.figure(figsize=(12, 7))

region_unemployment.sort_values().plot(kind="barh")

plt.title("Average Unemployment Rate by Region")
plt.xlabel("Unemployment Rate (%)")
plt.ylabel("Region")
plt.grid(axis="x")
plt.show()

```

# Final Findings

1. The unemployment rate changed over time and showed noticeable fluctuations.

2. The COVID-19 period had a significant effect on unemployment levels.

3. Unemployment rates were different across different regions of India.

4. Rural and urban areas showed differences in their average unemployment rates.

5. The analysis helped identify the highest and lowest unemployment rates in the dataset.

6. The visualizations made it easier to understand unemployment trends and regional differences.
